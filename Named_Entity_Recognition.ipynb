<h2 align='center'>NLP Tutorial: Named Entity Recognition (NER)</h2>

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")
for ent in doc.ents:
    print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))
```

```python
doc2 = nlp("Taminder is my roommate. We both go to the gym every morning. Prabhnoor is my another friend known for his singing. We all three are friends.")
for ent in doc2.ents:
    print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))
```

```python
from spacy import displacy
list = [doc, doc2]
displacy.render(list, style="ent")
```

```python
from spacy.tokens import Span
doc3_text = "Prabhnoor, Taminder and Tejinder are friends."
doc3 = nlp(doc3_text) # Process the string to create a Doc object
s1 = Span(doc3, 0, 1, label="PERSON")
s2 = Span(doc3, 2, 3, label="PERSON")
s3 = Span(doc3, 4, 5, label="PERSON") # Corrected index for Tejinder

doc3.set_ents([s1, s2, s3]) # Pass the list of spans to set_ents
```

```python
for ent in doc3.ents:
    print(ent.text, " | ", ent.label_)
```
